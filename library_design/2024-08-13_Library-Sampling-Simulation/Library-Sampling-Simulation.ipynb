```python
import numpy as np
from scipy.stats import multivariate_hypergeom
import matplotlib.pyplot as plt
```

```python
M = 1e10 # Number of DNA molecules (of marbles)
n = int(120000) # Number of colonies (marbles drawn)
k = 12000 # Number of variants (colors)
m = np.ones(int(k)) * M / k # Number of DNA molecules per variant (marbles per color)
m = m.astype(int)
x = np.ones(int(k)) * n / k # Number of colonies per variant (marbles per color)
x = x.astype(int)
n
# multivariate_hypergeom.pmf(, m=m, n = n)
```
