# Make multiple plots

```python
def plot_all_points_postinduction(df, fig, axs):
    axs[0,0].scatter( df.loc["major_axis_length"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=3, c = 'k', label='Post')
    axs[0,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=3, c = 'k', label='Post')
    axs[0,2].scatter( df.loc["minor_axis_length"]["Value"],\
            df.loc['Nucleoid Minor Axis Length Mean']["Value"],s=3, c = 'k', label='Post')
    axs[1,0].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["major_axis_length"]["Value"],s=3, c = 'k', label='Post')
    axs[1,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=3, c = 'k', label='Post')
    axs[1,2].scatter( df.loc["Total Nucleoid Area"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=3, c = 'k', label='Post')
    
    
    
def plot_all_points_preinduction(df, fig, axs):
    
    axs[0,0].scatter(df["major_axis_length"],\
                df["Total Nucleoid Area"],s=3, label = 'Pre')
    axs[0,1].scatter(df["Number of Nucleoids"],\
                df["Total Nucleoid Area"],s=3, label = 'Pre')
    axs[0,2].scatter(df["minor_axis_length"],\
                df['Nucleoid Minor Axis Length Mean'],s=3, label = 'Pre')
    axs[1,0].scatter(df["Number of Nucleoids"],\
                df["major_axis_length"],s=3, label = 'Pre')
    axs[1,1].scatter(df["Number of Nucleoids"],\
                df["minor_axis_length"],s=3, label = 'Pre')
    axs[1,2].scatter(df["Total Nucleoid Area"],\
                df["minor_axis_length"],s=3, label = 'Pre')
    
    
def plot_subset_postinduction_contains(df, query, fig, axs):
    df = df[df["Gene"].str.contains(query)]
    axs[0,0].scatter( df.loc["major_axis_length"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=30, label=query) 
    axs[0,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=30, label=query)
    axs[1,0].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["major_axis_length"]["Value"],s=30, label=query)
    axs[1,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=30, label=query)
    axs[1,2].scatter( df.loc["Total Nucleoid Area"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=30, label=query)
    
def plot_subset_postinduction_isin(df, query, label, fig, axs):
    df = df[df["Gene"].isin(query)]
    axs[0,0].scatter( df.loc["major_axis_length"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=30, label=label)
    axs[0,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["Total Nucleoid Area"]["Value"],s=30, label=label)
    axs[1,0].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["major_axis_length"]["Value"],s=30, label=label)
    axs[1,1].scatter( df.loc["Number of Nucleoids"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=30, label=label)
    axs[1,2].scatter( df.loc["Total Nucleoid Area"]["Value"],\
            df.loc["minor_axis_length"]["Value"],s=30, label=label)

def plot_subset_postinduction_contains_all(df, query_list, fig, axs):
    for query in query_list:
        plot_subset_postinduction_contains(df, query, fig, axs)
        
def plot_subset_postinduction_isin_all(df, query_list, label_list, fig, axs):
    for i, query in enumerate(query_list):
        plot_subset_postinduction_isin(df, query, label_list[i], fig, axs)
        
def plot_everything(estimator, list_contains, list_isin, list_isin_names):
    sns.set(font_scale = 1.2)
    sns.set_style("whitegrid", {'axes.grid' : False})
    # plt.style.use('dark_background')
    # sns.set_style("whitegrid")

    fig, axs = plt.subplots(nrows=2, ncols=3, figsize=(12,8))
    df_agg_estimator = df_agg.loc[estimator]

    plot_all_points_postinduction(df_agg_estimator, fig, axs)
    plot_all_points_preinduction(df_pre_agg, fig, axs)
    plot_subset_postinduction_contains_all(df_agg_estimator, list_contains, fig, axs)
    plot_subset_postinduction_isin_all(df_agg_estimator, list_isin,list_isin_names, fig, axs)

    axs[0,0].set_xlabel("Major axis length ($\mu$m)")
    axs[0,0].set_ylabel("Total Nucleoid Area ($\mu$m$^2$)")

    axs[0,1].set_xlabel("Number of Nucleoids")
    axs[0,1].set_ylabel("Total Nucleoid Area ($\mu$m$^2$)")

    axs[1,0].set_xlabel("Number of Nucleoids")
    axs[1,0].set_ylabel("Major axis length ($\mu$m)")

    axs[1,1].set_xlabel("Number of Nucleoids")
    axs[1,1].set_ylabel("Minor axis length ($\mu$m)")

    axs[1,2].set_xlabel("Total Nucleoid Area ($\mu$m$^2$)")
    axs[1,2].set_ylabel("Minor axis length ($\mu$m)")

    axs[0,0].legend(fontsize = 8)

    # handles, labels = axs[0,0].get_legend_handles_labels()
    # fig.legend(handles, labels, loc='upper center')

    fig.tight_layout()
```
