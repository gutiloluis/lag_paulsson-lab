```python
df_trench = pd.read_pickle("/home/lag36/scratch/lag36/2022-12-21_lDE25_Run_1/2023-01-03_lDE25_Steady_State_Timepoints_df_Trench_Estimators.pkl")
df_agg = pd.read_pickle("/home/lag36/scratch/lag36/2022-12-21_lDE25_Run_1/2023-01-03_lDE25_Steady_State_Timepoints_df_Estimators.pkl")

df_trench_pre = df_trench.loc['Pre', 'Mean']
df_trench_post = df_trench.loc["Post", 'Mean']

df_pre_agg = df_trench_pre.groupby("TargetID").agg({"major_axis_length": 'mean',
                                           'minor_axis_length': 'mean',
                                           'YFP mean_intensity': 'mean',
                                           'Number of Nucleoids': 'mean',
                                           'Total Nucleoid Area': 'mean',
                                           'oDEPool7_id': 'mean',
                                           'sgRNA': lambda x:x.iloc[0],
                                           'EcoWG1_id': 'mean',
                                           'Gene': lambda x:x.iloc[0],
                                           'N Mismatch': lambda x:x.iloc[0],
                                           'Category': lambda x:x.iloc[0]})
```

```python
ribostalk_df = df_agg[df_agg["Gene"].isin(["rplL","rplJ"])]
df_agg_fts = df_agg[df_agg["Gene"].str.contains("fts")]
muk_df = df_agg[df_agg["Gene"].str.contains("muk")]
par_df = df_agg[df_agg["Gene"].str.contains("par")]
dnaA_df = df_agg[df_agg["Gene"].str.contains("dnaA")]
rne_df = df_agg[df_agg["Gene"].isin(["rne"])]
rplK_df = df_agg[df_agg["Gene"].isin(["rplK"])]
```

```python
plt.scatter(df_agg.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            df_agg.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=3, label = 'Post')

plt.scatter(df_pre_agg["Number of Nucleoids"],\
            df_pre_agg["Total Nucleoid Area"],s=3, label = 'Pre')

plt.scatter(ribostalk_df.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            ribostalk_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'ribosome stalk')

plt.scatter(df_agg_fts.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            df_agg_fts.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'fts')

plt.scatter(muk_df.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            muk_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'muk')

plt.scatter(par_df.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            par_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'par')

plt.scatter(dnaA_df.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            dnaA_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'dnaA')

plt.scatter(rne_df.loc["Mean (Robust)","Number of Nucleoids"]["Value"],\
            rne_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label= 'rne')

plt.legend()
plt.xlabel('Number of Nucleoids')
plt.ylabel('Total Nucleoid Area')
```

```python
plt.scatter(df_agg.loc["Mean (Robust)","major_axis_length"]["Value"],\
            df_agg.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=3, label='Post')

plt.scatter(df_pre_agg["major_axis_length"],\
            df_pre_agg["Total Nucleoid Area"],s=3, label = 'Pre')

plt.scatter(df_agg_fts.loc["Mean (Robust)","major_axis_length"]["Value"],\
            df_agg_fts.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label='fts genes')

plt.scatter(rplK_df.loc["Mean (Robust)","major_axis_length"]["Value"],\
            rplK_df.loc["Mean (Robust)","Total Nucleoid Area"]["Value"],s=10, label='rplK')

plt.legend()
plt.xlabel('Major axis length')
plt.ylabel('Total Nucleoid Area')
plt.savefig('MajorAxisLength_vs_TotalNucArea.png', dpi = 500)
```
