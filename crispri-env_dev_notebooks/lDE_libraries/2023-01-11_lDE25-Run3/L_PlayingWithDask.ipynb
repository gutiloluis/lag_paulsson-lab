```python
# Set up a bigger example random timeseries
ddf = dask.datasets.timeseries(start="2000-01-01", end="2003-01-01", freq = '1h')

# Create the Pandas dataframe
df = ddf.compute()

# Time raising everything to the power of 2 using Dask
t1 = datetime.now()
a =  ddf.assign(x=ddf.x**2)
print(datetime.now() - t1)

# Time the same using Pandas
t1 = datetime.now()
b = df.assign(x=df.x**2)
print(datetime.now() - t1)
```
