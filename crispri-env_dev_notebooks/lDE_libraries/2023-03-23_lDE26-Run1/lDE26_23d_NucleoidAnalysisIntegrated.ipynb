```python
def load_dfs(path_preinduction, path_steady):
    df_pre = pd.read_pickle(path_preinduction)
    df_pre = df_pre[~df_pre['Gene'].isna()]
    
    df_post = pd.read_pickle(path_steady)
    df_post = df_post[~df_post['Gene'].isna()]
    return df_pre, df_post
```
