# Holoviews Tutorial

```python
import numpy as np
# import matplotlib.pyplot as plt
# import dask.dataframe as dd
import pandas as pd
import holoviews as hv
import panel as pn

hv.extension('bokeh')
pn.extension('tabulator')
```

## 1. Introduction

```python
# Simple curve
xs = [i for i in range(-10,11)]
ys = [100 - i**2 for i in xs]
simple_curve = hv.Curve((xs,ys))
simple_curve
```

```python
# Textual representation
print(simple_curve)
simple_curve.data
```

```python
hv.Scatter((xs,ys)) + hv.Area((xs,ys))
```

```python
# Annotating data
# kdims: Key dimension (coordinates, dependent variable)
# vdims: Value dimension (independent variable)
trajectory = hv.Curve((xs,ys), kdims=['distance'], vdims=['height'])
trajectory
```

```python
trajectory.vdims, trajectory.kdims
```

```python
# I can cast between Elements
trajectory.to(hv.Scatter) + hv.Scatter(trajectory) + hv.Curve(hv.Scatter(trajectory))
```

### Turning arrays into elements

```python
x = np.linspace(0, 10, 500)
y = np.linspace(0, 10, 500)

xx, yy = np.meshgrid(x, y)
arr = np.sin(xx)*np.cos(yy)
image = hv.Image(arr, kdims=['logitude', 'latitude'])
# Image bc it is mapping from two dimensions
image
```

### Selecting columns from tables to make elements

```python
economic_data = pd.read_csv('./Global Economy Indicators.csv')
economic_data.columns = economic_data.columns.str.strip()
# Remove starting and trailing spaces from all values
economic_data = economic_data.apply(lambda x: x.str.strip() if x.dtype == 'object' else x)
economic_data.tail()
# Remove spaces from all column names
```

```python
economic_data.columns
```

```python
US_data = economic_data[economic_data['Country'] == 'United States']
```

```python
# Declaring kdims and vdimes selects columns of the dataframe
growth_curve = hv.Curve(US_data, kdims=['Year'], vdims=['Gross Domestic Product (GDP)'])
growth_curve + hv.Curve(US_data, kdims=['Year'], vdims=['Gross National Income(GNI) in USD'])
```

### Dimension labels

```python
gdp_growth = growth_curve.redim.label(**{'Gross Domestic Product (GDP)':'GDP', 'Year':'Time'})
# If variable name does not have spaces, it can be used as function argument name
gdp_growth 
```

```python
gdp_growth.kdims
```

```python
gdp_growth.redim.unit(GDP='%', Time='year')
```

### Composing elements together
#### Layout

```python
layout = trajectory + hv.Scatter(trajectory) + hv.Area(trajectory) + hv.Spikes(trajectory)
layout.cols(2)
# The + operator creates layouts
```

```python
print(layout)
```

```python
layout.Area.I + layout.Spikes.I
```

```python
# We can label at the level of the element (not at the level of the dimension)
# Think about the group as two-level indexing
cannonball = trajectory.relabel('Cannonball', group='Trajectory')
integral = hv.Area(trajectory).relabel('Filled', group='Trajectory')
labelled_layout = cannonball + integral
labelled_layout
```

```python
# Think about the group as two-level indexing
labelled_layout.Trajectory.Filled + labelled_layout.Trajectory.Cannonball
# Equivalently in dict format
labelled_layout['Trajectory']['Filled'] + labelled_layout['Trajectory']['Cannonball']
```

#### Overlay
Overlay elements on a single plot

```python
trajectory*hv.Spikes(trajectory)
```

```python
labelled_layout['Trajectory']['Filled']*hv.Spikes(trajectory)
```

```python
# Clone keep semantic info while enabling us to change the data
# It is a new instance of the element, but it keeps the same metadata
tennis_ball = cannonball.clone((xs, 0.5*np.array(ys)), label='Tennis ball')
cannonball + tennis_ball + (cannonball*tennis_ball)
# Notice that the label shows up
```

### Slicing

```python
cannonball.kdims
```

```python
full_trajectory = cannonball.redim.label(distance = 'Horizontal distance', height = 'Vertical height')
ascending = full_trajectory[-10:1].relabel('Ascending') # Can select by index
descending = cannonball.select(distance=(0,11.)).relabel('Descending') # Or by value (even continuous) # Here it helps to have short dimension names, and keep longer names for labels
ascending*descending
```
